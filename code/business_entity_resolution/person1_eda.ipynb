# Business Entity Resolution - Person 1 EDA
This notebook loads the uploaded challenge data, checks schema and quality, and summarizes the training data before candidate generation.

```python
import pandas as pd
from pathlib import Path

base = Path('student_resource/dataset')
train_dir = base / 'train'
test_dir = base / 'test'

source1 = pd.read_csv(train_dir / 'train_source1.tsv', sep='\t')
source2 = pd.read_csv(train_dir / 'train_source2.tsv', sep='\t')
source3 = pd.read_csv(train_dir / 'train_source3.tsv', sep='\t')
gt = pd.read_csv(train_dir / 'train_ground_truth.tsv', sep='\t')

print('train_source1', source1.shape)
print('train_source2', source2.shape)
print('train_source3', source3.shape)
print('ground_truth', gt.shape)
print(source1.head(2).to_dict(orient='records'))
print(gt.head(2).to_dict(orient='records'))
```

```python
for name, df in [('source1', source1), ('source2', source2), ('source3', source3)]:
    print(name, 'missing', df.isna().sum().to_dict())
    print(name, 'duplicate_rows', int(df.duplicated().sum()))
    print(name, 'country_counts', df['country'].fillna('missing').value_counts().to_dict())
```

```python
gt['matched_count'] = gt['matched_entity_ids'].fillna('').str.split(',').map(lambda xs: len([x for x in xs if x.strip()]))
print(gt['matched_count'].value_counts().sort_index().to_dict())
print('entities_with_multiple_matches', int((gt['matched_count'] > 1).sum()))
print('entities_with_zero_matches', int((gt['matched_count'] == 0).sum()))
```
